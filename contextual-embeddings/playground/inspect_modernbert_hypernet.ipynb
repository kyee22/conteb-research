# Inspect ModernBERT HyperNet checkpoint

Load the saved hypernetwork model and print its parameter counts and module structure.

```python
from pathlib import Path
import sys
import torch

# Make the local contextual-embeddings package importable.
project_dir = Path.cwd()
if (project_dir / 'contextual_embeddings').is_dir():
    package_dir = project_dir
elif (project_dir / 'contextual-embeddings' / 'contextual_embeddings').is_dir():
    package_dir = project_dir / 'contextual-embeddings'
else:
    package_dir = Path('/root/kyx/conteb-research/contextual-embeddings')
sys.path.insert(0, str(package_dir))

from contextual_embeddings.models.hyper_context_model import HyperContextEmbeddingModel

checkpoint_candidates = [
    package_dir.parent / 'conteb' / 'checkpoints' / 'test' / 'modernbert-hypernet',
    package_dir / 'conteb' / 'checkpoints' / 'test' / 'modernbert-hypernet',
    Path('/root/kyx/conteb-research/conteb/checkpoints/test/modernbert-hypernet'),
]
checkpoint_path = next((path for path in checkpoint_candidates if path.is_dir()), None)
if checkpoint_path is None:
    raise FileNotFoundError('Could not locate modernbert-hypernet checkpoint')

model = HyperContextEmbeddingModel.from_checkpoint(str(checkpoint_path))

total_parameters = sum(parameter.numel() for parameter in model.parameters()) / 1_000_000
trainable_parameters = sum(parameter.numel() for parameter in model.parameters() if parameter.requires_grad) / 1_000_000
print(f'Checkpoint: {checkpoint_path}')
print(f'Total parameters: {total_parameters:,} M')
print(f'Trainable parameters: {trainable_parameters:,} M')
print('\nModel structure:')
print(model)
```

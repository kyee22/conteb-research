```python
from sentence_transformers import SentenceTransformer
from transformers import AutoModel
import sentence_transformers
import transformers


print(f"sentence_transformers version: {sentence_transformers.__version__}")
print(f"transformers version: {transformers.__version__}")
```

```python
modernbert = SentenceTransformer('lightonai/modernbert-embed-large')
pplx = SentenceTransformer('perplexity-ai/pplx-embed-v1-0.6B', trust_remote_code=True)
```

```python
sentences = ["hello, nju!", "this is a longer sentence! this is a longer sentence!"]
for encoder in [modernbert, pplx]:
    tokenizer = encoder.tokenizer
    ids = tokenizer.encode(sentences, add_special_tokens=True)
    print('=' * 20)
    print(f"Special token ID: {tokenizer.special_tokens_map}")
    print(f"SEP token ID: {tokenizer.sep_token_id}")
    print(f"pad token ID: {tokenizer.pad_token_id}")
    print([tokenizer.decode(id) for id in ids[0]])
    print([tokenizer.decode(id) for id in ids[1]])

print(modernbert.encode(sentences))
```

```python
for encoder in [modernbert, pplx]:
    ids1 = encoder.tokenizer.encode([encoder.tokenizer.sep_token.join(sentences)])
    ids2 = [encoder.tokenizer.encode(sentences[0])]\
         + [encoder.tokenizer.sep_token_id]\
         + [encoder.tokenizer.encode(sentences[1])]
    print(ids1)
    print(ids2)
```
